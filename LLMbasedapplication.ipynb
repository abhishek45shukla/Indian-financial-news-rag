```python
import pandas as pd 
df = pd.read_csv("IndianFinancialNews.csv")
print("Shape:", df.shape)
print(df.head())
```

```python
df = df.drop(columns=["Unnamed: 0"])
df["Description"] = df["Description"].fillna("")
df["Title"] = df["Title"].astype(str).str.strip()
df["Description"] = df["Description"].astype(str).str.strip()
df["Date"] = df["Date"].astype(str).str.strip()
```

```python
from langchain_core.documents import Document
documents = []

for _, row in df.iterrows():

    content = f"Title: {row['Title']}\nDescription: {row['Description']}"

    metadata = {
        "date": row["Date"],
        "title": row["Title"]
    }

    documents.append(
        Document(
            page_content=content,
            metadata=metadata
        )
    )
```

```python
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

print("Local embedding model loaded successfully!")
```

```python
from langchain_chroma import Chroma

persist_directory = "./vectorstore/chroma_db"

vectorstore = Chroma.from_documents(
    documents=chunks,
    embedding=embeddings,
    persist_directory=persist_directory,
    collection_name="indian_financial_news"
)

print("Vector database created successfully!")
print("Number of vectors:", vectorstore._collection.count())
```

```python
print("Number of vectors:", vectorstore._collection.count())
```

```python
retriever = vectorstore.as_retriever(
    search_type="similarity",
    search_kwargs={"k": 5}
)

print("Retriever created successfully!")
```

```python
query = "What are the latest developments in Indian banking?"

results = retriever.invoke(query)

for i, doc in enumerate(results, 1):
    print("=" * 80)
    print(f"RESULT {i}")
    print("=" * 80)
    print("Title:", doc.metadata.get("title"))
    print("Date:", doc.metadata.get("date"))
    print("\nContent:")
    print(doc.page_content)
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_template("""
You are an AI financial news research assistant.

Answer the user's question using ONLY the provided financial news context.

If the context does not contain enough information to answer the question,
clearly say that the information is not available in the retrieved news.

Do not invent facts, dates, companies, numbers, or events.

When possible, mention the relevant news title and date.

Context:
{context}

Question:
{question}

Answer:
""")

print("RAG prompt created successfully!")
```

```python
def rag_query(question, k=5):
    # Retrieve relevant news articles
    retrieved_docs = retriever.invoke(question)
    retrieved_docs = retrieved_docs[:k]

    # Create context
    context = "\n\n".join(
        [
            f"Title: {doc.metadata.get('title')}\n"
            f"Date: {doc.metadata.get('date')}\n"
            f"Content: {doc.page_content}"
            for doc in retrieved_docs
        ]
    )

    # Create prompt
    formatted_prompt = prompt.invoke({
        "context": context,
        "question": question
    })

    # Ask Gemini
    response = llm.invoke(formatted_prompt)

    # Extract clean text
    answer = response.text

    return answer, retrieved_docs


print("Updated RAG function successfully!")
```

```python
question = "What are the latest developments in Indian banking?"

answer, sources = rag_query(question)

print("ANSWER")
print("=" * 80)
print(answer)

print("\n\nSOURCES")
print("=" * 80)

for i, doc in enumerate(sources, 1):
    print(f"{i}. {doc.metadata.get('title')} ({doc.metadata.get('date')})")
```
